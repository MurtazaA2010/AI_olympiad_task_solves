```python
from sklearn.preprocessing import OrdinalEncoder
f_en = OrdinalEncoder().fit(train_ds[["food"]])
train_ds["food"]=f_en.transform(train_ds[["food"]])
```

```python
l_en = OrdinalEncoder().fit(train_ds[["light"]])
train_ds["light"]=l_en.transform(train_ds[["light"]])
```

```python
train_ds["food"].value_counts(), len(train_ds)
```

```python
from catboost import CatBoostClassifier
from xgboost import XGBClassifier

cat = CatBoostClassifier(task_type = "GPU", n_estimators=2000,learning_rate =0.05, class_weights = [2627/len(train_ds), 4118/len(train_ds),5702/len(train_ds), 11553/len(train_ds)]).fit(train_text, train_ds["food"])
```

```python
from xgboost import XGBClassifier
xgb = XGBClassifier( n_estimators=2000, learning_rate =0.05 ,device = "gpu").fit(train_text, train_ds["light"])
```

```python
food_pred= cat.predict(test_text)
l_pred= xgb.predict(test_text)
food_pred= f_en.inverse_transform(food_pred)
l_pred=l_en.inverse_transform(l_pred.reshape(-1,1))
```
